# Липсинк + сборка клипа «А представь, мы станем грибами»

**Что делает ноутбук.** Скачивает твои 26 клипов и песню с Диска, режет их на 25 слотов по 10 секунд (лишние куски, где в кадре видна раскадровка, отброшены), **прогоняет каждый слот с лицом через LatentSync** — то есть рот подгоняется под настоящий голос из песни, причём под тот именно кусок, который звучит в этом месте таймлайна, — и собирает финал **1920×1080** с плавными переходами и полным звуком.

**Как запустить:** меню `Среда выполнения` → `Сменить тип среды выполнения` → **T4 GPU** → `Среда выполнения` → `Выполнить всё`. Дальше ноутбук всё сделает сам и положит готовый файл на твой Диск.

**Сколько идёт:** установка ~5 минут, липсинк ~20–30 минут, сборка ~5 минут.

Ноутбук сам определит видеопамять: на T4 (16 ГБ) включит LatentSync 1.5, на L4/A100 (24–40 ГБ) — версию 1.6 (она резче).


```python
#@title 1. Проверка видеокарты
import torch, subprocess
print(subprocess.run(['nvidia-smi'], capture_output=True, text=True).stdout[:600] or 'GPU не найден!')
if not torch.cuda.is_available():
    raise SystemExit('Нет GPU. Меню «Среда выполнения» → «Сменить тип среды выполнения» → T4 GPU, потом запусти ячейку снова.')
print('Готово:', torch.cuda.get_device_name(0), round(torch.cuda.get_device_properties(0).total_memory/1e9, 1), 'ГБ')
```

```python
#@title 2. Установка LatentSync и весов (5 минут)
import os, subprocess, torch
os.chdir('/content')
subprocess.run(['git','clone','--quiet','--depth','1','https://github.com/bytedance/LatentSync.git'], check=False)
os.chdir('/content/LatentSync')
subprocess.run(['apt-get','-qq','-y','install','libgl1','libglib2.0-0'], capture_output=True)
subprocess.run(['pip','-q','install','gdown','imageio-ffmpeg'], check=True)

vram = torch.cuda.get_device_properties(0).total_memory/1e9
VER = '1.6' if vram >= 18 else '1.5'
CFG = 'configs/unet/stage2_512.yaml' if VER == '1.6' else 'configs/unet/stage2.yaml'
print(f'Видеопамять {vram:.1f} ГБ → LatentSync {VER}')

lines = [l for l in open('requirements.txt').read().splitlines()
         if l.strip() and not l.lower().startswith(('torch','xformers','insightface'))]
open('/content/req.txt','w').write('\n'.join(lines))
subprocess.run(['pip','-q','install','-r','/content/req.txt'], check=False)

from huggingface_hub import hf_hub_download
os.makedirs('checkpoints', exist_ok=True)
for f in ['whisper/tiny.pt', 'latentsync_unet.pt']:
    hf_hub_download(f'ByteDance/LatentSync-{VER}', f, local_dir='checkpoints')
print('Веса на месте:', os.listdir('checkpoints'))
```

```python
#@title 5. Липсинк по слотам, где есть лицо (20–30 минут на T4)
import os, shutil, subprocess
FACE = set([1, 4, 6, 7, 8, 9, 13, 14, 15, 16, 18, 19, 20, 21, 22, 23, 24, 25])
for i, s in enumerate(EDL):
    src, aud = f'{SEG}/s{i:02d}.mp4', f'{SEG}/a{i:02d}.wav'
    out = f'{SEG}/sync{i:02d}.mp4'
    if s['sheet'] not in FACE:
        shutil.copy(src, out); print(f'{i+1:02d} без лица — как есть'); continue
    base = ['python','-m','scripts.inference','--unet_config_path',CFG,
            '--inference_ckpt_path','checkpoints/latentsync_unet.pt',
            '--inference_steps','20','--guidance_scale','1.5',
            '--video_path',src,'--audio_path',aud,'--video_out_path',out]
    ok = False
    for extra in (['--enable_deepcache'], []):
        r = subprocess.run(base + extra, cwd='/content/LatentSync', capture_output=True, text=True)
        if os.path.exists(out) and os.path.getsize(out) > 10000:
            ok = True; break
        print(f'{i+1:02d} повтор без deepcache…')
    if not ok:
        shutil.copy(src, out); print(f'{i+1:02d} НЕ удалось — оставил оригинал')
    else:
        print(f'{i+1:02d} липсинк готов')
print('липсинк завершён')
```

```python
#@title 6. Сборка 1920×1080 и загрузка на Диск
import os, subprocess, shutil, imageio_ffmpeg
FF = shutil.which('ffmpeg') or imageio_ffmpeg.get_ffmpeg_exe()
N = len(EDL); DISS = 0.30
T = [0.0]
for s in EDL: T.append(T[-1] + s['dur'])
OUT = '/content/КЛИП_Грибы_1080p_липсинк.mp4'

# нормализуем каждый слот в 1920x1080 / 24 fps
for i in range(N):
    subprocess.run([FF,'-y','-i',f'{SEG}/sync{i:02d}.mp4','-vf',
        'fps=24,scale=1920:1080:force_original_aspect_ratio=increase,crop=1920:1080,setsar=1,'
        'tpad=stop_mode=clone:stop_duration=1.0','-an','-c:v','libx264','-preset','fast','-crf','15',
        f'{SEG}/f{i:02d}.mp4'], check=True)

for ci,(a,b) in enumerate([(1,13),(13,26)]):
    ins = []
    for k in range(a-1,b-1): ins += ['-i', f'{SEG}/f{k:02d}.mp4']
    fc, prev = [], '[0:v]'
    for k in range(a, b-1):
        off = (T[k] - T[a-1]) - DISS
        fc.append(f'{prev}[{k-a+1}:v]xfade=transition=fade:duration={DISS}:offset={off:.3f}[v{k}]')
        prev = f'[v{k}]'
    # мягкая «шторка» на стыке двух блоков и плавное затемнение в финале
    dur_blk = round(T[b-1]-T[a-1], 3)
    if ci == 0:
        fc.append(f'{prev}fade=t=out:st={dur_blk-0.18:.3f}:d=0.18,format=yuv420p[v]')
    else:
        fc.append(f'{prev}fade=t=in:st=0:d=0.18,fade=t=out:st={dur_blk-1.6:.3f}:d=1.6,format=yuv420p[v]')
    subprocess.run([FF,'-v','error','-y']+ins+['-filter_complex',';'.join(fc),'-map','[v]',
        '-c:v','libx264','-preset','slow','-crf','17','-pix_fmt','yuv420p','-r','24','-an',
        '-t',f'{dur_blk:.3f}',f'/content/prt{ci}.mp4'], check=True)
    print('блок', ci+1, 'готов')

open('/content/prt.txt','w').write("file '/content/prt0.mp4'\nfile '/content/prt1.mp4'\n")
subprocess.run([FF,'-v','error','-y','-f','concat','-safe','0','-i','/content/prt.txt','-i','/content/song.mp3',
    '-map','0:v','-map','1:a','-c:v','copy','-c:a','aac','-b:a','320k','-ar','48000',
    '-af','afade=t=out:st=247.2:d=2.2','-movflags','+faststart','-t','249.480',OUT], check=True)
print('собрано:', round(os.path.getsize(OUT)/1e6,1), 'МБ')

dst = '/content/drive/MyDrive'
if os.path.isdir(dst + '/Грибы'): dst += '/Грибы'
shutil.copy(OUT, dst)
print('залито на Диск:', dst)
```

## Если что-то пойдёт не так

- **«Нет GPU»** — включи T4 в меню «Сменить тип среды выполнения».
- **Не хватает памяти при липсинке** — это версия 1.6 на маленькой карте; ноутбук сам переключится на 1.5, если видеопамяти меньше 18 ГБ. Если ошибка всё же появилась, запусти ячейку 5 повторно: она продолжит с того места.
- **Отдельный слот не удался** — он останется без липсинка, остальные соберутся нормально, в логе будет видно какой.
